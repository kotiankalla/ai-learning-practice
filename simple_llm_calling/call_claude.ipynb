```python
from dotenv import load_dotenv
from langchain_anthropic import ChatAnthropic

load_dotenv()
```

```python
llm = ChatAnthropic(
    model="claude-sonnet-4-5-20250929",
    max_tokens=200
)

response = llm.invoke(
    "Explain SQL Injection in one simple paragraph."
)

print(response.content)
```
