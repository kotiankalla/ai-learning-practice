```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()

```

```python
llm= ChatGoogleGenerativeAI(model="gemma-4-31b-it")
response = llm.invoke("How many moons does Jupiter have?")
print(response.text)
```

```python
response = llm.invoke([
    ["system", "your are a smart and helpful assistant that answers in short one line"],
    ["human", "how many moons does Jupiter have?"]
])
print(response.text)
```

```python
prompt = "Write a single tagline for a quirky coffee shop. Only output the tagline. nothing else."
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash",temperature=0)
response = llm.invoke(prompt)
print(response.text)
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash",temperature=2)
response = llm.invoke(prompt)
print(response.text)
```

```python
from langchain_groq import ChatGroq
groqllm = ChatGroq(
    model="qwen/qwen3-32b", 
    reasoning_format="parsed"
)
messages=[
    (
        "system", "your are a smart and helpful assistant that gives a short answer to a question"
    ),
    (
        "human", "How many moons does Jupiter have?"
    )
]
ai_msg = groqllm.invoke(messages)
print(ai_msg.content)
```

```python
import os
import requests

# Fetch all available models directly from Groq's endpoint
api_key = os.environ.get("GROQ_API_KEY")
url = "https://api.groq.com/openai/v1/models"
headers = {"Authorization": f"Bearer {api_key}"}

response = requests.get(url, headers=headers).json()

print("--- Your Active Groq Models ---")
for model in response.get("data", []):
    print(model["id"])
```
