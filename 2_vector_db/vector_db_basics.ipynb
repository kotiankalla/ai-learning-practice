```python
import chromadb

client = chromadb.Client()
```

```python
collection = client.create_collection("news_articles")
collection.add(
    ids = ["id1","id2","id3","id4"],
    documents=[
        "Apple is leading in a smart phone game with iPhone sales up by 35%",
        "Tesla booked a minor profit of 1 billion $ in Q2",
        "Apples are high in fiber, Vitamin C and various anioxidants",
        "SpaceX got NASA contract worth 10 billion $"
    ]
)
```

```python
# fetch documents and embeddings
data = collection.get(
    include = ["documents", "embeddings", "metadatas"]
)

for i, emb in enumerate(data["embeddings"]):
    print(f"\n Document {i+1}")
    print("Text: ", data["documents"][i])
    print("Embedding Lenght:", len(emb))
    print("First 10 Values: ", emb[:10])
```

```python
results = collection.query(
    query_texts = "This is query related to Elon Musk",
    n_results = 2
)
print(results)
```

```python
results = collection.query(
    query_texts="I love iPad",
    n_results=4
)
print(results)
```
